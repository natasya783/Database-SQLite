```python
7. DATABASE SQLITE (Bobot 10%)
# =====================================================================
DB_NAME = "care_motor.db"

def koneksi_db():
    """Membuka koneksi ke database SQLite."""
    return sqlite3.connect(DB_NAME)

def inisialisasi_database():
    """Membuat tabel dan relasi database jika belum tersedia."""
    with koneksi_db() as conn:
        cur = conn.cursor()
        cur.execute("""
            CREATE TABLE IF NOT EXISTS user (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                nama TEXT NOT NULL,
                username TEXT NOT NULL UNIQUE,
                password TEXT NOT NULL
            )
        """)
        cur.execute("""
            CREATE TABLE IF NOT EXISTS kendaraan (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                user_id INTEGER NOT NULL,
                nomor_polisi TEXT NOT NULL,
                merk TEXT NOT NULL,
                tipe TEXT NOT NULL,
                tahun INTEGER NOT NULL,
                kilometer_saat_ini INTEGER NOT NULL CHECK(kilometer_saat_ini >= 0),
                FOREIGN KEY(user_id) REFERENCES user(id)
            )
        """)
        cur.execute("""
            CREATE TABLE IF NOT EXISTS riwayat_servis (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                id_kendaraan INTEGER NOT NULL,
                jenis_servis TEXT NOT NULL,
                km_saat_servis INTEGER NOT NULL,
                tanggal_servis TEXT NOT NULL,
                biaya INTEGER NOT NULL CHECK(biaya >= 0),
                km_target_berikutnya INTEGER NOT NULL,
                catatan TEXT,
                FOREIGN KEY(id_kendaraan) REFERENCES kendaraan(id)
            )
        """)
        conn.commit()


# =====================================================================
# MENU UTAMA & EKSEKUSI PROGRAM
# =====================================================================
def menu_pengguna(user):
    """Menjalankan menu utama sampai pengguna memilih logout."""
    while True:
        print("\n========== CARE MOTOR ==========")
        print("1. Tambah Kendaraan")
        print("2. Lihat Kendaraan")
        print("3. Update Kilometer")
        print("4. Tambah Riwayat Servis")
        print("5. Jadwal Servis Berikutnya")
        print("6. Notifikasi Servis")
        print("7. Laporan Kendaraan")
        print("8. Riwayat Servis Kendaraan")
        print("9. Export Riwayat ke CSV (Bonus)")
        print("10. Logout")
        pilihan = input("Pilih menu: ").strip()

        if pilihan == "1":
            tambah_kendaraan(user["id"])
        elif pilihan == "2":
            lihat_kendaraan(user["id"])
        elif pilihan == "3":
            update_kilometer(user["id"])
        elif pilihan == "4":
            tambah_riwayat_servis(user["id"])
        elif pilihan == "5":
            jadwal_servis_berikutnya(user["id"])
        elif pilihan == "6":
            notifikasi_servis(user["id"])
        elif pilihan == "7":
            laporan_kendaraan(user["id"])
        elif pilihan == "8":
            tampilkan_riwayat(user["id"])
        elif pilihan == "9":
            export_csv(user["id"])
        elif pilihan == "10":
            print("Logout berhasil.")
            break
        else:
            print("Pilihan tidak valid. Silakan coba lagi.")

if __name__ == "__main__":
    inisialisasi_database()
    menu_awal()
```
